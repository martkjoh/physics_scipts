```julia
# Only for backend C (WGLMakie). Run once per kernel session.
# `exportable=true, offline=true` inlines the Bonito JS so JupyterLab can render
# the canvas without any frontend extension. Keep this cell's output visible —
# clearing it removes the JS context and subsequent plots will be blank.
# Page(exportable=true, offline=true)
```

```julia
xs = range(-2, 2; length=60)
ys = range(-2, 2; length=60)
zs = [x^2 + y^2 for x in xs, y in ys]

fig = Figure(size=(700, 600))
ax  = Axis3(fig[1, 1]; xlabel="x", ylabel="y", zlabel="z = x² + y²")
surface!(ax, xs, ys, zs; colormap=:viridis)
Colorbar(fig[1, 2]; colormap=:viridis, limits=extrema(zs), label="z")

show_fig(fig)
```
