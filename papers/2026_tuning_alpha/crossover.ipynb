```python
import numpy as np
import matplotlib.pyplot as plt

from numpy import exp, abs, log as ln
from matplotlib import cm
from matplotlib.animation import FuncAnimation as FA

plt.rc("font", family="serif", size=16)
plt.rc("mathtext", fontset="cm")
# 
plt.rc("lines", lw=2)
```

```python
b0 = 1.1
l1 = 2 - 2/5
l2 = 2
a0 = 1
r0 = 1
K0 = 1
g0 = b0/K0 * r0/a0

def a(l): return a0*(g0 *exp(l1 * l) + (1-g0)* exp(l2 * l))
def lc(g0) : return 1 / (l1 - l2) * ln(abs((1 - g0)/ (g0)))

print(lc)
```

```python
fig, ax = plt.subplots(figsize=(3, 4))
g0 = np.linspace(-1., 2.1, 1000)
ax.plot(g0, lc(g0), label='$\\ell_c$')
ax.plot([np.min(g0), np.max(g0)], [0, 0], 'k-', lw=2, alpha=.5)
ax.plot([1/2, 1/2], [-5, 5], 'k--', lw=1)
ax.plot([1, 1], [-5, 5], 'k--', lw=1)

ax.set_xlabel("$g_0$")
ax.set_ylabel("$\\ell_c$")
ax.set_ylim(-5, 5)
ax.legend()
fig.savefig("crossover/RG_time.pdf", bbox_inches='tight')
```

```python
from matplotlib.ticker import LogFormatter

l = np.linspace(0, 14, 500) 

size = 4
b0s = [0.5, 0.9, 1.1]
g0s = [b0 * (r0/a0) for b0 in b0s]
N = len(g0s)
fig, axs = plt.subplots(1, N, figsize=(4*N, 4/1.2),sharey=True)
fig.subplots_adjust(wspace=0.0)

for i, g0 in enumerate(g0s):
    ax = axs[i]

    sign = a(l) > 0
    ax.plot(l*sign, a(l)*sign, 'tab:blue', lw=4., label="$+\\alpha_0(\\ell)$ ", alpha=.4, zorder=10)
    sign = a(l) < 0
    ax.plot(l*sign, -a(l)*sign, 'tab:green', lw=4., label="$-\\alpha_0(\\ell)$ ", alpha=.4, zorder=10)
    ax.plot(l, np.abs(g0)*r0*np.exp(l1*l), 'r--', label="$e^{\\lambda_r \\ell}$")
    ax.plot(l, a0*abs(1-g0)*np.exp(l2*l), 'k--', label="$e^{\\lambda_\\sigma \\ell}$")

    ax.plot([lc(g0), lc(g0)], [a0*abs(1-g0),a0*abs(1-g0) * exp( l2 * l[-1])], '--', color="gray", lw=1.)

    ax.set_yscale('log')
    ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f'{int(np.log10(x))}'))

    ax.set_xlabel("$\\ell$")
    if lc(g0)>0: ax.text(lc(g0) + .5, 10, "$\\ell_\\times$")


    if i==1:
        ax.text(2, 2e2, "$ \\sim e^{\\lambda_r \\ell} $", rotation=35)
        ax.text(7, 2e6, "$\\sim e^{\\lambda_\\sigma \\ell} $", rotation=45)

    ax.set_title("$K \\alpha_0/\\beta_0 r"+"={g0:3.4}$".format(g0=g0))
    ax.set_xlim(0,12)

axs[0].legend()
axs[0].set_ylabel("$\\log_{10} \\alpha_0(\\ell)$")

axs[0].set_ylim(1e0,9e9)
plt.show()
fig.savefig("crossover/crossover.pdf", bbox_inches='tight')
```

```python
from matplotlib.ticker import LogFormatter

l = np.linspace(0, 8, 500) 

fig, ax = plt.subplots(figsize=(6, 3.5))
w = 2/5
b = .5

for i,d in enumerate([.4, .1, -1, -.2, -1.6]):
    s = np.sign(d*b)
    ax.plot(l, np.abs(b + d*np.exp(w*l)), color=cm.viridis(i/4), lw=4., label="$\\Delta = {},\\, b= {}$".format(d, b), alpha=.4, zorder=10)
    lnr = 1 / w * np.log( (1 - np.sign(d*b)* np.abs(b)) / np.abs(d))
    ax.plot(lnr, 1, 'rx')

ax.plot(lnr, 1, 'rx', label="$\\ell_\\mathrm{nr}$")


ax.plot(l, l/l, '--k')
ax.set_yscale('log')
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f'{int(np.log10(x))}'))
ax.set_xlabel("$\\ell$")
ax.set_ylabel("$\\log_{10} | b + \\Delta e^{|\\omega| \\ell} |$")

ax.set_ylim(0.1,800)
plt.legend(fontsize=12)
plt.show()
fig.savefig("crossover/crossover2.pdf", bbox_inches='tight')
```
